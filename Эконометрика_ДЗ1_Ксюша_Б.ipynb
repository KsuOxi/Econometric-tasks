```python
import matplotlib.gridspec as gridspec
fig = plt.figure(figsize=(12, 5))
gs = gridspec.GridSpec(1, 2, width_ratios=[2, 1])

ax1 = fig.add_subplot(gs[0, 0])
sns.histplot(data=df, x='Previous Purchases', bins=20, kde=True, ax=ax1)
ax1.set_title('Распределение предыдущих покупок')
ax1.set_xlabel('Previous Purchases')
ax1.set_ylabel('Count')

ax2 = fig.add_subplot(gs[0, 1])
df['Frequency of Purchases'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    ax=ax2
)
ax2.set_ylabel('')
ax2.set_title('Распределение частоты покупок')

plt.tight_layout()
plt.show()
```

Частота покупок не имеет доминирующих сценариев, покупатели распределены равномерно между режимами (ежемесячно, ежегодно, ежеквартально и т.д.).

Вывод: Можно заметить, что обувь и верхняя одежда в целом покупаются довольно редко. Чаще всего покупаются одежда и аксессуары. Одежда обновляется у людей каждые месяца (высокое влияние моды), тогда как аксессуары чаще всего покупаются ежеквартально, что в свою очередь связано со сменой сезонов (летом - солнцезащитные очки, зимой и осенью - шарфы и т.д.)

Одежда очень редко покупается с частотой раз в две недели, поскольку мода не меняется настолько быстро и срок жизни одежды намного выше чем 2 недели. Аксессуары же реже всего покупаются еженедельно.

Различие средних сумм покупок для разных категорий товаров (Category) и для разных сезонов покупки (Season)

```python
df.pivot_table(
    values="Purchase Amount (USD)",
    index="Category",
    columns="Season",
    aggfunc="mean"
)
```

Вывод: самый высокий средний чек по все категориям наблюдается осенью. Такая тенденция может быть обоснована подготовкой покупателей к более холодным сезонам, а также с различными распродажами. Весной люди в среднем тратят меньше чем в предыдущие сезоны, что может быть связано с сокращением денежных средств после зимних празников. Самая дорогостоящая категория осенью - обувь. Обувь в целом стоит дороже, чем одежда, и цена на нее в сезон (то есть осенью) держится на высоком уровне в связи с высоким спросом.

```python
df.groupby("Category")["Purchase Amount (USD)"].mean().sort_values(ascending=False)
```

6. Какие особенности имеют покупки с наибольшей стоимостью, включая категорию товара, сезон покупки, способ оплаты и наличие скидок?

```python
q9 = df["Purchase Amount (USD)"].quantile(0.9)
expensive = df[df["Purchase Amount (USD)"] >= q9]

expensive.groupby("Category")["Purchase Amount (USD)"].mean().sort_values(ascending=False)
#выявляем дорогие покупки и смотрим средний чек по категориям
```

```python
expensive['Category'].value_counts(normalize=True) * 100
#сразу видим, что категории распределены неравномерно
```

```python
expensive['Discount Applied'].value_counts(normalize=True) * 100
```

```python
expensive['Season'].value_counts(normalize=True) * 100
```

```python
expensive['Payment Method'].value_counts(normalize=True) * 100
```

Вывод: Особенностями товаров с высокой стоимостью является отсутствиее скидок на сами товары в 61% покупок, так как высокая стоимость подчеркивает эксклюзивность одежды, ее качество, статусность для покупателей.

Большая часть таких покупок была оплачена дебетовыми картами, что говорит о том, что ЦА дорогой одежды - люди с высоким доходом, у которых даже такие значительные суммы лежат на обычном счете.

Больше всего дорогих покупок приходятся на категории одежды и аксессуаров, а большая их часть (около трети) была совершена осенью. Это может быть связано с тем, что осенью возникает необходимоть одеваться теплее и многослойнее (куртка+блузка/свитер/толстовка+брюки+шарф+шапка+перчатки).
